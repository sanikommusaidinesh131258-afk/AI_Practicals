```python
print("Hello! I am Robo.")
print("Today we will build a cleaning robot together. 🤖")
print(2 + 3, "is the answer to 2 + 3")

room = ["D", "C", "D", "D", "C"]

def show_room(room):
    picture = ""
    for spot in room:
        if spot == "D":
            picture += "💩 "
        else:
            picture += "✨ "
    print(picture)

print("Our room right now:")
show_room(room)

def clean_spot(spot):
    if spot == "D":
        return "C"
    else:
        return "C"


result = clean_spot("D")
print("The robot looked at a dirty spot 💩 and made it:", result,
      "(C means clean ✨)")

print("BEFORE - the dirty room:")
show_room(room)


for i in range(len(room)):
    room[i] = clean_spot(room[i])
    print("After cleaning spot number", i + 1, ":")
    show_room(room)

print()
print("AFTER - all done! ✨🤖")
```

```python
print("Creating Basic AI Chatbot")
name = input("Enter your name ")
print(f"Hello, {name}!")

user_problem = input("Enter your problem: ")

#def some issue in chart bot
responses = {
    "hi": "Hello! How can I help you today?",
    "hello": "Hi there! What can I do for you?",
    "help": "I can assist you with account issues, tracking orders, or basic FAQs",
    "pricing": "Our plans start at 10/month. View details on our website",
    "training": "Your account is in training mode please be patiences",
    "bye": "Goodbye! Have a great day!"
}

def get_bot_response(user_message):
    # Convert to lowercase and strip punctuation to match keys
    clean_message = user_message.lower().strip(" ")

    return responses.get(clean_message, "I am sorry, I didn't quite catch that.")

# Example loop to test it
while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break
    print(f"Bot: {get_bot_response(user_input)}")

print("Your Response has been submitted")
print("Thank you for your patience.")
```

```python
print("email spam dection")
words = ["money","loan","congurats","bank details","discount","otp"]
email_text = input("Enter the email text: ")

is_spam = False
for word in words:
    if word in email_text.lower():
        is_spam = True
        break

if is_spam:
    print("This email is likely spam.")
else:
    print("This email is likely not spam.")
```
