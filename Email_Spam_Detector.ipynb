```python
print("email spam dection")
words = ["money","loan","congurats","bank details","discount","otp"]
email_text = input("Enter the email text: ")

is_spam = False
for word in words:
    if word in email_text.lower():
        is_spam = True
        break

if is_spam:
    print("This email is likely spam.")
else:
    print("This email is likely not spam.")
```
