```python
print("Creating Basic AI Chatbot")
name = input("Enter your name ")
print(f"Hello, {name}!")

user_problem = input("Enter your problem: ")

#def some issue in chart bot
responses = {
    "hi": "Hello! How can I help you today?",
    "hello": "Hi there! What can I do for you?",
    "help": "I can assist you with account issues, tracking orders, or basic FAQs",
    "pricing": "Our plans start at 10/month. View details on our website",
    "training": "Your account is in training mode please be patiences",
    "bye": "Goodbye! Have a great day!"
}

def get_bot_response(user_message):
    # Convert to lowercase and strip punctuation to match keys
    clean_message = user_message.lower().strip(" ")

    return responses.get(clean_message, "I am sorry, I didn't quite catch that.")

# Example loop to test it
while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break
    print(f"Bot: {get_bot_response(user_input)}")

print("Your Response has been submitted")
print("Thank you for your patience.")
```
